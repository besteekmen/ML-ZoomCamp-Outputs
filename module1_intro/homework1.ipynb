#### Q7. Median value of horsepower

```python
med_old = df["horsepower"].median()
print("Old median:", med_old)

mode = df["horsepower"].mode()[0]
print("Most frequent horsepower:", mode)

df["horsepower"] = df["horsepower"].fillna(mode)
med_new = df["horsepower"].median()
print("New median:", med_new)

if med_new < med_old:
    print("Yes, it decreased")
elif med_new > med_old:
    print("Yes, it increased")
else:
    print("No")
```

#### Q8. Sum of weights

```python
df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).values
```

```python
import numpy as np

X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).values
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)
print("Sum of all the elements of the result:", w.sum())
```
